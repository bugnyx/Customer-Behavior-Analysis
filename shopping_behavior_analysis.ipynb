```python
import pandas as pd

df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.head()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.replace(' ', '_').str.lower()
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
#Create a new column age group
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
#accurate way to divide into age groups
import numpy as np
bins = [18, 35, 50, 65, np.inf]
labels = ["Young Adult", "Adult", "Middle-Aged", "Senior"]

df["age_group"] = pd.cut(
    df["age"], bins=bins, labels=labels, right=True, include_lowest=True
)
```

```python
df['age_group'].value_counts()
```

```python
df[['age', 'age_group']].head(10)
```

```python
#create column purchase_frequency_days

frequency_mapping = {
	'Weekly': 7,
	'Fortnightly': 14,
	'Bi-Weekly': 14,
	'Monthly': 30,
	'Quarterly': 90,
	'Every 3 Months': 90,
	'Annually': 365,
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
#do we need both discount_applied and promo_code_used columns?

df[['discount_applied', 'promo_code_used']].head(10)
```

```python
#both columns are identical, we can drop one of them. Let's drop the promo_code_used column.
df = df.drop(columns=['promo_code_used'])
```
